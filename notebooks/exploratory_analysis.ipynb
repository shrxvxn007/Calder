# CALDER saved-result exploration
This notebook reads existing benchmark outputs only. Run `calder synthetic --config configs/synthetic.yaml` first. Synthetic outcomes are simulated. Biological outputs, when present, are retrospective public-data replays; they are not prospective laboratory validation.
Do not tune final RPE1 hyperparameters using this notebook.

```python
from pathlib import Path
import json
import pandas as pd
from IPython.display import Image, display

root = Path.cwd() if (Path.cwd() / 'pyproject.toml').exists() else Path.cwd().parent
runs = sorted(p for p in (root / 'results').iterdir() if (p / 'experiment_manifest.json').exists())
if not runs:
    raise RuntimeError('No saved runs; execute the documented benchmark CLI first.')
run_dir = runs[-1]  # Set explicitly to reproduce a particular analysis.
manifest = json.loads((run_dir / 'experiment_manifest.json').read_text())
display(manifest)
print(run_dir)
```

```python
for filename in ['results.csv', 'seed_level_results.csv', 'calibration.csv', 'ablations.csv']:
    path = run_dir / filename
    if path.exists():
        print(filename)
        table = pd.read_csv(path)
        display(table.head(20))
```

```python
figures = sorted(run_dir.rglob('*.png'))
for figure in figures:
    print(figure.relative_to(run_dir))
    display(Image(filename=str(figure)))
```
